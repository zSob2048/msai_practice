```python
print(maxunicode) # unicode 기준 문자 개수
print(chr(65)) # : 65번째 unicode 문자
print(category("A")) # unicode category

special = []
for i in range(maxunicode):
    # print(chr(i), category(chr(i))) 
    if category(chr(i)).startswith("P"): # 특수문자
        special.append(chr(i))
```

```python
f = open("C:\MS_AI_25\practice\py/bd\pg22914.txt", "r", encoding="utf-8")

wc = {}

for line in f.readlines():
    for s in special:
        line = line.replace(s, "")
    line = line.replace("\n", "").split(" ")

    for word in line:
        if word in wc:
            wc[word] += 1
        else:
            wc[word] = 1
f.close()
```

```python
sent = "문장을 작성해 주세요..."

wt = word_tokenize(sent)
print(wt)
st = sent_tokenize(sent)
print(st)
# 단어단위 / 문장단위

ensent = "sentence on eng..."

# 불용어(광사, 조사, 대명사, 접속사 등...  )
sw = stopwords.words("english")
print(sw)

ewt = word_tokenize(ensent)
for w in ewt:
    w = w.lower()
    if w not in sw:
        print(w)


ps = PorterStemmer() # 어간 추출기

words = "variable variety variation"
wwt = word_tokenize(words)
for w in wwt:
    print(w, ps.stem(w))

    
wnl = WordNetLemmatizer() # 표제어 추출기(동사 원형)

bwords = "am are is"
bwt = word_tokenize(bwords)
for w in bwt:
    print(w, wnl.lemmatize(w, wordnet.VERB))


# 품사확인
pt = pos_tag(ewt)
print(pt)
# [('sentence', 'NN'), ('on', 'IN'), ('eng', 'NN'), ('...', ':')] : tuple

```
