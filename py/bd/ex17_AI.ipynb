```python
import numpy as np

featuree = np.array([[80,20],[95,5],[10,90],[90,10],[5,95]])
labell = np.array(["A","A","B","A","B"])

from sklearn.neighbors import KNeighborsClassifier

knc = KNeighborsClassifier(3) # k 입력
knc.fit(featuree, labell) # 학습
```

```python
import mlflow

mlflow.start_run()

import numpy as np

featuree = np.array([[80,20],[95,5],[10,90],[90,10],[5,95]])
labell = np.array(["A","A","B","A","B"])

from sklearn.neighbors import KNeighborsClassifier

knc = KNeighborsClassifier(3) # k 입력
knc.fit(featuree, labell) # 학습

x = 35
y = 65
z = np.array([[x,y]])


print(knc.predict(z))

mlflow.sklearn.save_model(sk_model=knc,path="./zKNNModel") # 모델 저장 경로

mlflow.end_run()

```
