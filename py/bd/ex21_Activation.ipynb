activation(활성화 함수)
    Linear Activation : 사실상 의미없지 선형만큼 좃밥짱코엽고간단한함수가없으니까
    Unit Step Ftn : 1 (x>0)
                    0 (x<0)
                    미분값 균일성 + 0점 불연속으로 GD 불가
    sigmoid : 1/ (1+e^-x) 좋은함수지연
            미분값의 상계 1/4
            >> 적용될수록 G값이 줄어.. 적용된 층이 많아질수록 0에 가까워짐(vanishing gradient)
    ReLU(Restricted Linear Unit) : max(0,x)
        // 입력값이 음수인경우 죽은 뉴런이 됨(Dying ReLU)
        근데 이것도 0에서 연속이다 뿐이지 미분은 안되는데
