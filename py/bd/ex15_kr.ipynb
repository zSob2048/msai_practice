```python
from konlpy.tag import Okt

from jamo.jamo import h2j, j2hcj
from unicode import join_jamos

from wordcloud.wordcloud import WordCloud
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

fontF = "C:\Windows\Fonts\gulim.ttc"
fontN = fm.FontProperties(fname=fontF, size= 10).get_name()
plt.rc("font", family=fontN)
plt.rcParams["axes.unicode_minus"]=False
```

```python
o = Okt()

sent = "다람쥐 헌 챗바퀴에 어쩌고저쩌고"

print(o.normalize(sent)) # 정규화(정리)
print(o.phrases(sent)) # 어구 추출
print(o.morphs(sent)) # 형태소 분석
print(o.morphs(sent,stem=True)) # +어간추출
print(o.pos(sent)) # 형태소+품사 tuple
print(o.pos(sent, join=True)) # 형태소/품사 형태로 출력
print(o.pos(sent, stem=True))

```

```python
t = "합성"
print(h2j(t))
# 초성/ 중성/ 종성 구분해서 분리 // VSCode에서는 합성상태 출력
print(j2hcj(h2j(t))) # 초성 중성 종성 구분없는 상태로 변환
print(join_jamos(j2hcj(h2j(t))))
```
